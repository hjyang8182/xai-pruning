```python
import json
import os
from src import config
from src.visualise import plot_lambda_gate_sweep

DATASETS = ['cifar100', 'cub', 'food101', 'oxford_pet']

for dataset in DATASETS:
    path = os.path.join(config.DATA_PATH, dataset, 'lambda_gate_sweep.json')
    with open(path) as f:
        d = json.load(f)

    plot_lambda_gate_sweep(
        d['lambda_gates'], d['cea'], d['open_gates'], dataset,
        accs_std=d['cea_std'], n_open_std=d['open_gates_std'],
        baseline_acc=d['baseline_cea'], baseline_acc_std=d['baseline_cea_std'],
        test_accs=d['test_accuracy'], test_accs_std=d['test_accuracy_std'],
        baseline_test_acc=d['baseline_test_accuracy'], baseline_test_acc_std=d['baseline_test_accuracy_std'],
        val_metric_name='CEA',
    )
    print(f'{dataset}: plotted')
```

```python
import json
import glob
import os
from src import config
from src.visualise import plot_concept_ablation

DATASETS = ['cifar100', 'cub', 'food101', 'oxford_pet']

for dataset in DATASETS:
    paths = sorted(glob.glob(os.path.join(config.DATA_PATH, dataset, 'concept_ablation_*.json')))
    path = paths[-1]
    with open(path) as f:
        d = json.load(f)

    plot_concept_ablation(
        d['fractions'], d['open_acc_mean'], d['closed_acc_mean'],
        d['open_acc_std'], d['closed_acc_std'], dataset,
        random_mean=d.get('random_acc_mean'), random_std=d.get('random_acc_std'),
    )
    print(f'{dataset}: plotted from {os.path.basename(path)}')
```

```python
import json
import glob
import os
from src import config
from src.visualise import plot_concept_ablation_grid

DATASETS = ['cifar100', 'cub', 'food101', 'oxford_pet']

results_by_dataset = {}
for dataset in DATASETS:
    paths = sorted(glob.glob(os.path.join(config.DATA_PATH, dataset, 'concept_ablation_*.json')))
    path = paths[-1]
    with open(path) as f:
        d = json.load(f)
    results_by_dataset[dataset] = d
    print(f'{dataset}: loaded from {os.path.basename(path)}')

path = plot_concept_ablation_grid(results_by_dataset, dataset_order=DATASETS)
print(f'Saved combined grid to {path}')
```
